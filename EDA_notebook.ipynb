# EDA Notebook
This notebook performs quick EDA on the cleaned sales dataset.

```python
import pandas as pd
sales = pd.read_csv('data_sales_cleaned.csv', parse_dates=['OrderDate'])
sales.head()
```

```python
# Monthly sales
sales['YearMonth'] = sales['OrderDate'].dt.to_period('M')
monthly = sales.groupby('YearMonth').TotalSales.sum().reset_index()
monthly.head()
```
